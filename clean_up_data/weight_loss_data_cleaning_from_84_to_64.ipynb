Drop the columns of comments.
Six-month time period: Whether the exam took place Nov–Apr or May–Oct (survey scheduling).

Body Measures Component Status Code: Whether the body measurements were completed, partial, or not done.

Other comments.

```python
import pandas as pd

df = pd.read_csv('nhanes_84_col.csv')

admin_cols = [
    "Six-month time period",
    "Body Measures Component Status Code",
    "Weight Comment",
    "Standing Height Comment",
    "Upper Leg Length Comment",
    "Upper Arm Length Comment",
    "Arm Circumference Comment",
    "Waist Circumference Comment",
    "Hip Circumference Comment",
]
df = df.drop(columns=admin_cols)
print(f"Shape after dropping columns: {df.shape}")
```

## Drop the constant or empty columns.

Data release cycle: Always 12, the code for the 2021–2023 cycle.

Tried to lose weight in past year: Always 1 (yes), because we already filtered to people who tried.

Questionnaire Mode Flag: Same interview mode for everyone left in the adult sample.

Cigarettes smoked in entire life: Completely blank in the adult sample.

Age first smoked whole cigarette: Completely blank in the adult sample.

```python
constant_empty_cols = [
    "Data release cycle",
    "Tried to lose weight in past year",
    "Questionnaire Mode Flag",
    "Cigarettes smoked in entire life",
    "Age first smoked whole cigarette",
]
df = df.drop(columns=constant_empty_cols)
```

```python
print(f"Shape after dropping columns: {df.shape}")
```

```python
df.to_csv('nhannes_70_col.csv', index=False)
print(f"Dataframe successfully saved to 'nhannes_70_col.csv' with shape {df.shape}")
```

## Drop 5 leakage columns that have the measurement outcomes.

Weight (kg): Measured at the exam, after the year of weight change. It’s essentially the current weight, which is half of outcome formula.

Body Mass Index (kg/m**2): Calculated from that same current measured weight.

Waist Circumference (cm), Hip Circumference (cm), Arm Circumference (cm): Shrinks or grows with weight change, so it reflects the result rather than predicting it.

```python
leakage_cols = [
    "Weight (kg)",
    "Body Mass Index (kg/m**2)",
    "Waist Circumference (cm)",
    "Hip Circumference (cm)",
    "Arm Circumference (cm)",
]
df = df.drop(columns=leakage_cols)
```

```python
print(f"Shape after dropping columns: {df.shape}")
```

```python
df.to_csv('nhanes_64_col.csv', index=False)
print(f"Dataframe successfully saved to 'nhanes_64_col.csv' with shape {df.shape}")
```
