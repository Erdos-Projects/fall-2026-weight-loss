## Data Exploration
Let's start by loading the dataset and examining its basic properties.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/nhanes_model_ready.csv')

# Display the first few rows
display(df.head())
```

```python
# Check the shape and data types
print("Dataset Information:")
df.info()
```

```python
# Summary statistics for numerical columns
display(df.describe())
```

### Visualizations
Let's visualize the distributions of some key features and explore their relationships.

```python
# 1. Distribution of the target variable (lost_5pct)
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='lost_5pct')
plt.title('Distribution of 5% Weight Loss (Target)')
plt.xlabel('Lost 5% of Body Weight (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.show()
```

```python
# 2. Histograms of Age and BMI
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df['age'], bins=30, kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Age Distribution')
axes[0].set_xlabel('Age')

sns.histplot(df['bmi_1yr_ago'], bins=30, kde=True, ax=axes[1], color='salmon')
axes[1].set_title('BMI (1 Year Ago) Distribution')
axes[1].set_xlabel('BMI')

plt.tight_layout()
plt.show()
```

### Missing Data Analysis
Let's calculate the percentage of missing data for each column in the dataset.

```python
# Calculate percentage of missing data
missing_percent = (df.isnull().sum() / len(df)) * 100

# Filter to show only columns with missing data and sort them
missing_percent = missing_percent[missing_percent > 0].sort_values(ascending=False)

# Display as a dataframe
missing_df = pd.DataFrame({'Missing Percentage (%)': missing_percent})
print("Columns with missing data:")
display(missing_df)
```

### Behavioral Features Correlation Heatmap
Let's check the correlations among the specific behavioral variables.

```python
behavioral_cols = [
    "mod_min_wk", "vig_min_wk", "sedentary_hr_day", "phq9_score",
    "drinks_per_week", "sleep_hr_avg", "smoking_status_Former",
    "smoking_status_Current", "main_meal_planner", "shares_meal_planning",
    "main_food_shopper", "shares_food_shopping"
]

# Calculate correlation matrix for behavioral features
behavioral_corr = df[behavioral_cols].corr()

# Plot the heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(behavioral_corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Heatmap of Behavioral Features')
plt.tight_layout()
plt.show()
```

### Covariates Correlation Heatmap
Let's visualize the correlations among the covariate variables (excluding the behavioral ones).

```python
labels = {
    "age": "Age (years)", "female": "Female", "education": "Education (1-5)",
    "income_poverty_ratio": "Income-to-poverty ratio", "household_size": "Household size",
    "us_born": "US-born", "veteran": "Veteran", "bmi_1yr_ago": "BMI 1 year ago",
    "mod_min_wk": "Moderate activity (min/wk)", "vig_min_wk": "Vigorous activity (min/wk)",
    "sedentary_hr_day": "Sedentary (hr/day)", "phq9_score": "PHQ-9 depression score",
    "drinks_per_week": "Drinks per week", "sleep_hr_avg": "Sleep (hr/night)",
    "main_meal_planner": "Main meal planner", "shares_meal_planning": "Shares meal planning",
    "main_food_shopper": "Main food shopper", "shares_food_shopping": "Shares food shopping",
    "smoking_status_Former": "Former smoker", "smoking_status_Current": "Current smoker",
    "race_eth_Mexican American": "Mexican American", "race_eth_Other Hispanic": "Other Hispanic",
    "race_eth_NH Black": "NH Black", "race_eth_NH Asian": "NH Asian",
    "race_eth_Other/Multiracial": "Other/Multiracial",
    "marital_Widowed/divorced/separated": "Widowed/divorced/separated",
    "marital_Never married": "Never married",
}

behavioral = ["mod_min_wk", "vig_min_wk", "sedentary_hr_day", "phq9_score", "drinks_per_week",
              "sleep_hr_avg", "smoking_status_Former", "smoking_status_Current",
              "main_meal_planner", "shares_meal_planning", "main_food_shopper", "shares_food_shopping"]

covariates = [c for c in labels if c not in behavioral]

# Calculate correlation matrix for covariates
covariate_corr = df[covariates].corr()

# Plot the heatmap
plt.figure(figsize=(16, 14))
sns.heatmap(covariate_corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Heatmap of Covariates')
plt.tight_layout()
plt.show()
```
