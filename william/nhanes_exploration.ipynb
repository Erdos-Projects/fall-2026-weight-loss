```python
# Check whether near-zero activity frequencies correspond to missing durations

print("Moderate activity:")
print(
    paq.loc[paq["PAD790Q"].between(0, 1e-50), "PAD800"]
    .value_counts(dropna=False)
)

print("\nVigorous activity:")
print(
    paq.loc[paq["PAD810Q"].between(0, 1e-50), "PAD820"]
    .value_counts(dropna=False)
)
```

```python
import numpy as np

# Preserve the original dataset
paq_clean = paq.copy()

# Clean moderate and vigorous activity frequencies
for col in ["PAD790Q", "PAD810Q"]:
    paq_clean[col] = paq_clean[col].replace(
        [7777, 9999], np.nan
    )

    # Correct near-zero SAS values to actual zero
    paq_clean.loc[
        paq_clean[col].between(0, 1e-50), col
    ] = 0

# Verify the cleaning
for col in ["PAD790Q", "PAD810Q"]:
    print(f"\n{col}:")
    print("Zero activity:", (paq_clean[col] == 0).sum())
    print("Missing:", paq_clean[col].isna().sum())
```

```python
# Examine special response codes for activity duration

for col in ["PAD800", "PAD820"]:
    print(f"\n{col}:")
    print("Refused (7777):", (paq_clean[col] == 7777).sum())
    print("Don't know (9999):", (paq_clean[col] == 9999).sum())
    print("Missing:", paq_clean[col].isna().sum())
    print("Minimum:", paq_clean[col].min())
    print("Maximum:", paq_clean[col].max())
```

```python
# Clean physical activity duration variables

for col in ["PAD800", "PAD820"]:
    paq_clean[col] = paq_clean[col].replace(
        [7777, 9999], np.nan
    )

# Check the results
paq_clean[["PAD800", "PAD820"]].describe()
```

```python
# Convert frequency unit codes from bytes to strings
for col in ["PAD790U", "PAD810U"]:
    paq_clean[col] = paq_clean[col].str.decode("utf-8")

# Convert reported activity frequencies to weekly equivalents
def sessions_per_week(frequency, unit):
    if pd.isna(frequency):
        return np.nan
    if frequency == 0:
        return 0
    if unit == "D":
        return frequency * 7
    if unit == "W":
        return frequency
    if unit == "M":
        return frequency * 12 / 52
    if unit == "Y":
        return frequency / 52
    return np.nan

paq_clean["moderate_sessions_week"] = paq_clean.apply(
    lambda row: sessions_per_week(row["PAD790Q"], row["PAD790U"]),
    axis=1
)

paq_clean["vigorous_sessions_week"] = paq_clean.apply(
    lambda row: sessions_per_week(row["PAD810Q"], row["PAD810U"]),
    axis=1
)

# Examine the new variables
paq_clean[
    ["moderate_sessions_week", "vigorous_sessions_week"]
].describe()
```

```python
# Examine participants reporting more than 14 activity occasions per week

print("Unusually high moderate activity:")
display(
    paq_clean.loc[
        paq_clean["moderate_sessions_week"] > 14,
        ["SEQN", "PAD790Q", "PAD790U",
         "PAD800", "moderate_sessions_week"]
    ].sort_values("moderate_sessions_week", ascending=False)
)

print("Unusually high vigorous activity:")
display(
    paq_clean.loc[
        paq_clean["vigorous_sessions_week"] > 14,
        ["SEQN", "PAD810Q", "PAD810U",
         "PAD820", "vigorous_sessions_week"]
    ].sort_values("vigorous_sessions_week", ascending=False)
)
```

```python
# Calculate estimated minutes of activity per week

paq_clean["moderate_min_week"] = (
    paq_clean["moderate_sessions_week"] * paq_clean["PAD800"]
)

paq_clean["vigorous_min_week"] = (
    paq_clean["vigorous_sessions_week"] * paq_clean["PAD820"]
)

# Participants reporting no activity should have zero minutes
paq_clean.loc[
    paq_clean["moderate_sessions_week"] == 0,
    "moderate_min_week"
] = 0

paq_clean.loc[
    paq_clean["vigorous_sessions_week"] == 0,
    "vigorous_min_week"
] = 0

# Examine the resulting distributions
paq_clean[
    ["moderate_min_week", "vigorous_min_week"]
].describe()
```

```python
# Calculate total leisure-time physical activity minutes per week

paq_clean["total_ltpa_min_week"] = paq_clean[
    ["moderate_min_week", "vigorous_min_week"]
].sum(axis=1, min_count=2)

# Examine the new variable
paq_clean["total_ltpa_min_week"].describe()
```

```python
# Merge Demographics, Weight History, and Physical Activity

nhanes = (
    demo
    .merge(whq_clean, on="SEQN", how="left", validate="one_to_one")
    .merge(paq_clean, on="SEQN", how="left", validate="one_to_one")
)

# Restrict to adults aged 20 or older
nhanes_adults = nhanes[nhanes["RIDAGEYR"] >= 20].copy()

# Examine the merged dataset
print("Merged dataset:", nhanes.shape)
print("Adult sample:", nhanes_adults.shape)
```

```python
# Examine missing data for our primary research variables

variables = [
    "WHD020",             # Current weight
    "WHD050",             # Weight one year ago
    "WHQ070",             # Weight-loss attempts
    "moderate_min_week",  # Moderate physical activity
    "vigorous_min_week",  # Vigorous physical activity
    "total_ltpa_min_week" # Total physical activity
]

missing_summary = pd.DataFrame({
    "Valid N": nhanes_adults[variables].notna().sum(),
    "Missing N": nhanes_adults[variables].isna().sum(),
    "Missing %": nhanes_adults[variables].isna().mean() * 100
})

missing_summary.round(2)
```

```python
# Calculate percentage weight change

nhanes_adults["pct_weight_change"] = (
    (nhanes_adults["WHD020"] - nhanes_adults["WHD050"])
    / nhanes_adults["WHD050"]
) * 100

# Examine descriptive statistics
nhanes_adults["pct_weight_change"].describe()
```

```python
# Identify participants with complete data for the primary analysis

primary_vars = [
    "pct_weight_change",
    "WHQ070",
    "total_ltpa_min_week",
    "RIDAGEYR",
    "RIAGENDR"
]

primary_sample = nhanes_adults.dropna(
    subset=primary_vars
).copy()

print("Total adults:", len(nhanes_adults))
print("Complete cases:", len(primary_sample))
print(
    "Percentage retained:",
    round(len(primary_sample) / len(nhanes_adults) * 100, 2)
)
```

## Revised Project Aim: Predictive Modeling

The primary aim of this project is to predict percentage weight change among U.S. adults using NHANES 2021–2023 data. Rather than focusing primarily on hypothesis testing, we will use supervised machine-learning methods to evaluate how accurately demographic, behavioral, and health-related characteristics predict weight change.

Potential predictors include weight-loss attempts, leisure-time physical activity, age, sex, sleep, alcohol consumption, smoking, depressive symptoms, and selected medical conditions.

Our approach will involve:

1. Exploratory Data Analysis: Examine distributions, missingness, relationships between variables, and potential outliers.
2. Feature Selection and Engineering: Identify relevant predictors, construct meaningful variables, and assess potential target leakage.
3. Baseline Modeling: Compare a dummy regression model with linear regression and tree-based models.
4. Model Evaluation: Use training/test splits and cross-validation to assess predictive performance using MAE, RMSE, and R².
5. Feature Importance: Explore which characteristics contribute most to predictive performance.

Because NHANES 2021–2023 is cross-sectional, our models will focus on retrospective prediction of reported weight change over the preceding year, rather than forecasting future weight loss. We will also consider the implications of NHANES's complex survey design when evaluating model generalizability.

The original hypothesis concerning leisure-time physical activity as a moderator of the association between weight-loss attempts and percentage weight change may be retained as a secondary exploratory question.

```python
# Create a copy of the Alcohol Use dataset
alq_clean = alq.copy()

# Correct near-zero SAS values to zero
alq_clean.loc[
    alq_clean["ALQ121"].between(0, 1e-50),
    "ALQ121"
] = 0

# Convert special missing-value codes
alq_clean["ALQ121"] = alq_clean["ALQ121"].replace(
    [77, 99], np.nan
)

alq_clean["ALQ130"] = alq_clean["ALQ130"].replace(
    [777, 999], np.nan
)

# Approximate drinking days per week
frequency_map = {
    0: 0,
    1: 7,
    2: 6,
    3: 3.5,
    4: 2,
    5: 1,
    6: 2.5 * 12 / 52,
    7: 12 / 52,
    8: 9 * 1 / 52,
    9: 4.5 / 52,
    10: 1.5 / 52
}

alq_clean["drinking_days_week"] = (
    alq_clean["ALQ121"].map(frequency_map)
)

# Assign zero to lifetime nondrinkers
alq_clean.loc[
    alq_clean["ALQ111"] == 2,
    "drinking_days_week"
] = 0

# Examine the new variable
alq_clean["drinking_days_week"].describe()
```

```python
# Calculate estimated average alcoholic drinks per week

alq_clean["drinks_per_week"] = (
    alq_clean["drinking_days_week"] * alq_clean["ALQ130"]
)

# Assign zero consumption to participants reporting no drinking
alq_clean.loc[
    alq_clean["drinking_days_week"] == 0,
    "drinks_per_week"
] = 0

# Examine descriptive statistics and missingness
print(alq_clean["drinks_per_week"].describe())

print(
    "\nMissing observations:",
    alq_clean["drinks_per_week"].isna().sum()
)
```


### Alcohol Consumption: Missingness and Feature Selection

We created `drinks_per_week`, an estimated measure of average weekly alcohol consumption, using reported drinking frequency and quantity.

Among 7,809 adults:
- 5,202 (66.62%) had valid alcohol-consumption data.
- 1,745 (22.35%) had no corresponding Alcohol Use questionnaire record.
- 862 (11.04%) had a questionnaire record but insufficient information to calculate weekly consumption.
- Overall missingness was 33.38%.

Feature selection decision: We will retain `drinks_per_week` as a candidate predictor but exclude it from our initial baseline model because of its relatively high missingness. We may evaluate its contribution in an expanded model using appropriate missing-data handling.

The reasons for absent questionnaire records require further investigation. Estimated weekly alcohol consumption also relies on approximations of categorical drinking frequencies.

This decision is consistent with Checkpoint 2's emphasis on evaluating missingness, feature relevance, and data quality before modeling.

```python
# Define candidate predictors
model_features = [
    "RIDAGEYR",
    "RIAGENDR",
    "education",
    "marital_status",
    "race_ethnicity",
    "poverty_ratio",
    "WHQ070",
    "baseline_bmi",
    "moderate_min_week",
    "vigorous_min_week",
    "sitting_min_day",
    "avg_sleep_hours",
    "smoking_status",
    "employment_status",
    "medical_condition_count",
    "hypertension_history",
    "high_cholesterol_history"
]

# Define the outcome
target = "pct_weight_change"

# Retain participants with a valid outcome
model_data = nhanes_adults.dropna(
    subset=[target]
).copy()

# Separate predictors (X) from outcome (y)
X = model_data[model_features].copy()

# Convert pandas missing values to NumPy NaN
X = X.astype(object).replace({pd.NA: np.nan})

y = model_data[target].copy()

# Verify dimensions
print("Modeling sample:", len(model_data))
print("Number of predictors:", X.shape[1])

print("\nMissing values by predictor:")
print(X.isna().sum())
```

```python
# Numerical predictors
numeric_features = [
    "RIDAGEYR",
    "poverty_ratio",
    "baseline_bmi",
    "moderate_min_week",
    "vigorous_min_week",
    "sitting_min_day",
    "avg_sleep_hours",
    "medical_condition_count"
]

# Categorical predictors
categorical_features = [
    "RIAGENDR",
    "education",
    "marital_status",
    "race_ethnicity",
    "WHQ070",
    "smoking_status",
    "employment_status",
    "hypertension_history",
    "high_cholesterol_history"
]

# Verify that every predictor is accounted for
print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
print("Total:", len(numeric_features) + len(categorical_features))

assert set(numeric_features + categorical_features) == set(X_train.columns)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical preprocessing
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine preprocessing steps
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
```
