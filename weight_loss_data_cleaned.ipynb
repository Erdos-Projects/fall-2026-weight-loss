## Merging these file together into a dataframe.

```python
import pandas as pd

# Load all processed CSV files
demo_df = pd.read_csv("DEMO_L.csv")
bmx_df = pd.read_csv("BMX_L.csv")
paq_df = pd.read_csv("PAQ_L.csv")
dbq_df = pd.read_csv("DBQ_L.csv")
slq_df = pd.read_csv("SLQ_L.csv")
smq_df = pd.read_csv("SMQ_L.csv")
alq_df = pd.read_csv("ALQ_L.csv")
dpq_df = pd.read_csv("DPQ_L.csv")
whq_df = pd.read_csv("WHQ_L.csv")

# List of DataFrames to merge on 'SEQN'
dataframes = [demo_df, bmx_df, paq_df, dbq_df, slq_df, smq_df, alq_df, dpq_df, whq_df]

# Perform sequential outer merge using SEQN to retain all records
df_merged = dataframes[0]
for next_df in dataframes[1:]:
    df_merged = pd.merge(df_merged, next_df, on="SEQN", how="outer")

# Save the final merged dataset to CSV
df_merged.to_csv("nhanes_merged_2021_2023.csv", index=False)

print(f"Merged Dataset Shape: {df_merged.shape}")
print(df_merged.head())
```

## Filter out people who have tried to lose weight in past year (column WHQ070 with 1 = Yes, 2 = No, 7/9 = refused/don’t know).

```python
import pandas as pd

# Load the merged dataset
df_merged = pd.read_csv("nhanes_merged_2021_2023.csv")

# Prior to filtering, let's look at the distribution of WHQ070
print("Distribution of WHQ070 before filtering:")
print(df_merged['WHQ070'].value_counts(dropna=False))

# Filter out people who have tried to lose weight (WHQ070 == 1)
# We keep those who did not try to lose weight (2), refused/don't know (7/9), or have missing values (NaN)
df_filtered = df_merged[df_merged['WHQ070'] != 1]

# Save the filtered dataset
df_filtered.to_csv("nhanes_filtered_weight_loss.csv", index=False)

print(f"\nOriginal dataset shape: {df_merged.shape}")
print(f"Filtered dataset shape: {df_filtered.shape}")
print(f"Successfully filtered out {len(df_merged) - len(df_filtered)} participants who tried to lose weight in the past year.")
```

## Change the columns to the respective meaning.

```python
import pandas as pd
import pyreadstat

# Load the dataset of people who tried to lose weight
df = pd.read_csv("nhanes_tried_to_lose_weight.csv")

# List of all xpt files used
xpt_files = [
    'DEMO_L.xpt', 'BMX_L.xpt', 'PAQ_L.xpt', 'DBQ_L.xpt',
    'SLQ_L.xpt', 'SMQ_L.xpt', 'ALQ_L.xpt', 'DPQ_L.xpt', 'WHQ_L.xpt'
]

# Create a dictionary to map column names to their meanings
column_mapping = {}

for file in xpt_files:
    try:
        _, meta = pyreadstat.read_xport(file, encoding="latin1")
        for col_name, col_label in zip(meta.column_names, meta.column_labels):
            column_mapping[col_name] = col_label
    except Exception as e:
        print(f"Error reading {file}: {e}")

# Rename the columns in the dataframe
df_renamed = df.rename(columns=column_mapping)

# Save the renamed dataset
df_renamed.to_csv("nhanes_tried_to_lose_weight_renamed.csv", index=False)

print("Successfully renamed columns. Here are the first few renamed columns:")
print(list(df_renamed.columns[:10]))
```

Pregnancy status at exam:
1 = Pregnant, 2 = Not pregnant, 3 = Cannot ascertain

```python
col = "Pregnancy status at exam"
print(df[col].value_counts(dropna=False))   # see how many are 1, 2, 3, or blank
```

```python
import numpy as np

preg = "Pregnancy status at exam"
age  = df["Age in years at screening"]
sex  = df["Gender"]                          # 1 = Male, 2 = Female
exam = df["Interview/Examination status"]    # 1 = interview only, 2 = interview + exam

# Put each person into one group (checked in this order)
group = np.select(
    [sex == 1,
     exam == 1,
     age < 20,
     age > 44],
    ["Men",
     "Women not examined",
     "Women under 20 (examined)",
     "Women 45+ (examined)"],
    default="Women 20-44 (examined)",
)

# Give the pregnancy codes readable names; blanks become "Blank"
status = df[preg].map({1: "Pregnant", 2: "Not pregnant", 3: "Can't tell"}).fillna("Blank")

# Use df.index to ensure proper alignment when creating the crosstab
table = pd.crosstab(
    pd.Series(group, index=df.index, name="Group"),
    pd.Series(status, name="Pregnancy status"),
    margins=True, margins_name="Total",
)

# Put rows and columns in a sensible order
row_order = ["Men", "Women 45+ (examined)", "Women under 20 (examined)",
             "Women not examined", "Women 20-44 (examined)", "Total"]
col_order = ["Pregnant", "Not pregnant", "Can't tell", "Blank", "Total"]
table = table.reindex(index=row_order, columns=col_order, fill_value=0)

table
```
