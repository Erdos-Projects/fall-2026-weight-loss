## Merging these file together into a dataframe.

```python
import pandas as pd

# Load all processed CSV files
demo_df = pd.read_csv("DEMO_L.csv")
bmx_df = pd.read_csv("BMX_L.csv")
paq_df = pd.read_csv("PAQ_L.csv")
dbq_df = pd.read_csv("DBQ_L.csv")
slq_df = pd.read_csv("SLQ_L.csv")
smq_df = pd.read_csv("SMQ_L.csv")
alq_df = pd.read_csv("ALQ_L.csv")
dpq_df = pd.read_csv("DPQ_L.csv")
whq_df = pd.read_csv("WHQ_L.csv")

# List of DataFrames to merge on 'SEQN'
dataframes = [demo_df, bmx_df, paq_df, dbq_df, slq_df, smq_df, alq_df, dpq_df, whq_df]

# Perform sequential outer merge using SEQN to retain all records
df_merged = dataframes[0]
for next_df in dataframes[1:]:
    df_merged = pd.merge(df_merged, next_df, on="SEQN", how="outer")

# Save the final merged dataset to CSV
df_merged.to_csv("nhanes_merged_2021_2023.csv", index=False)

print(f"Merged Dataset Shape: {df_merged.shape}")
print(df_merged.head())
```

## Filter out people who have tried to lose weight in past year (column WHQ070 with 1 = Yes, 2 = No, 7/9 = refused/don’t know).

```python
import pandas as pd

# Load the merged dataset
df_merged = pd.read_csv("nhanes_merged_2021_2023.csv")

# Prior to filtering, let's look at the distribution of WHQ070
print("Distribution of WHQ070 before filtering:")
print(df_merged['WHQ070'].value_counts(dropna=False))

# Filter out people who have tried to lose weight (WHQ070 == 1)
# We keep those who did not try to lose weight (2), refused/don't know (7/9), or have missing values (NaN)
df_filtered = df_merged[df_merged['WHQ070'] != 1]

# Save the filtered dataset
df_filtered.to_csv("nhanes_filtered_weight_loss.csv", index=False)

print(f"\nOriginal dataset shape: {df_merged.shape}")
print(f"Filtered dataset shape: {df_filtered.shape}")
print(f"Successfully filtered out {len(df_merged) - len(df_filtered)} participants who tried to lose weight in the past year.")
```

## Make dataframe of people who tried to lose weight in the past year.

```python
import pandas as pd

# Load the merged dataset
df_merged = pd.read_csv("nhanes_merged_2021_2023.csv")

# Filter to keep only people who tried to lose weight (WHQ070 == 1)
df_tried_to_lose = df_merged[df_merged['WHQ070'] == 1]

# Save the filtered dataset to CSV
df_tried_to_lose.to_csv("nhanes_tried_to_lose_weight.csv", index=False)

print(f"Original dataset shape: {df_merged.shape}")
print(f"Dataset shape of people who tried to lose weight: {df_tried_to_lose.shape}")
print(f"Successfully saved {len(df_tried_to_lose)} records to 'nhanes_tried_to_lose_weight.csv'.")
```

## Remove people who have 7777/9999 for current weight WHD020.

```python
import pandas as pd

# Load the dataset of people who tried to lose weight
df_tried_to_lose = pd.read_csv("nhanes_tried_to_lose_weight.csv")

# Prior to filtering, let's see how many people have 7777 or 9999 in WHD020
invalid_weights = df_tried_to_lose['WHD020'].isin([7777, 9999]).sum()
print(f"Number of participants with 7777 or 9999 in WHD020: {invalid_weights}")

# Filter out rows where WHD020 is 7777 or 9999
df_cleaned = df_tried_to_lose[~df_tried_to_lose['WHD020'].isin([7777, 9999])]

# Save the cleaned dataset to CSV
df_cleaned.to_csv("nhanes_tried_to_lose_weight.csv", index=False)

print(f"\nOriginal dataset shape: {df_tried_to_lose.shape}")
print(f"Cleaned dataset shape: {df_cleaned.shape}")
print(f"Successfully saved cleaned records back to 'nhanes_tried_to_lose_weight.csv'.")
```

## Inpect any other missing data.

```python
import pandas as pd

# Load the cleaned dataset
df = pd.read_csv("nhanes_tried_to_lose_weight.csv")

# Calculate missing values
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Combine into a summary DataFrame
missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Percentage (%)': missing_percentage
})

# Sort and display columns with missing data (excluding columns with 0 missing values)
missing_summary = missing_summary[missing_summary['Missing Values'] > 0].sort_values(by='Missing Values', ascending=False)

pd.set_option("display.max_rows", None)
print(f"Total columns with missing data: {len(missing_summary)} out of {df.shape[1]} columns\n")
print(missing_summary)
```

## Filter out the predictors that have less than 30\% missing data.

```python
import pandas as pd

# Load the dataset of people who tried to lose weight
df = pd.read_csv("nhanes_tried_to_lose_weight.csv")

# Calculate the percentage of missing values for each column
missing_percentage = df.isnull().mean() * 100

# Identify columns with less than 30% missing data
cols_to_keep = missing_percentage[missing_percentage < 30].index.tolist()

# Filter the dataframe
df_filtered_predictors = df[cols_to_keep]

# Save the filtered dataset to a new CSV file
df_filtered_predictors.to_csv("nhanes_predictors_under_30_percent_missing.csv", index=False)

print(f"Original dataset shape: {df.shape}")
print(f"Filtered dataset shape (retaining columns with < 30% missingness): {df_filtered_predictors.shape}")
print(f"Successfully removed {df.shape[1] - df_filtered_predictors.shape[1]} columns due to high missingness.")
```
